# Exercício 3 — Estatísticas futebolísticas

## Enunciado

Desenvolva um algoritmo em Python que receba a informação do resultado de 5 jogos entre Avaí e Figueirense e armazene em dicionários. Além disso, o programa deverá apresentar as seguintes informações:

1. Listagem de jogos.
   - Ex.: Jogo 1 (Avaí x Figueirense): 4 x 3
   - Jogo 2 (Avaí x Figueirense): 3 x 4
   - etc.
2. Quantidades de vitórias do Avaí.
3. Quantidades de vitórias do Figueirense.
4. Quantidade de empates.
5. Saldo de gols do Avaí.
6. Saldo de gols do Figueirense.

**Requisito:** as informações de saída deverão ser calculadas com base nos dados armazenados nos dicionários.

## Estratégia

Cada partida será representada por um dicionário com o número do jogo e os gols de cada equipe. Os cinco dicionários ficarão armazenados na lista `jogos`. Depois, um percurso pela lista apresentará as partidas e acumulará vitórias, empates e saldo de gols.

```python
# Entrada dos resultados: cada jogo será guardado em um dicionário.
jogos = []

for numero in range(1, 6):
    print(f'Jogo {numero} (Avaí x Figueirense)')
    gols_avai = int(input('Gols do Avaí: '))
    gols_figueirense = int(input('Gols do Figueirense: '))

    jogo = {
        'numero': numero,
        'avai': gols_avai,
        'figueirense': gols_figueirense
    }
    jogos.append(jogo)

# Inicializa os contadores e saldos.
vitorias_avai = 0
vitorias_figueirense = 0
empates = 0
saldo_avai = 0
saldo_figueirense = 0

# Calcula as estatísticas a partir dos dicionários armazenados.
for jogo in jogos:
    gols_avai = jogo['avai']
    gols_figueirense = jogo['figueirense']

    if gols_avai > gols_figueirense:
        vitorias_avai += 1
    elif gols_figueirense > gols_avai:
        vitorias_figueirense += 1
    else:
        empates += 1

    saldo_avai += gols_avai - gols_figueirense
    saldo_figueirense += gols_figueirense - gols_avai

print('\nListagem de jogos:')
for jogo in jogos:
    print(f"Jogo {jogo['numero']} (Avaí x Figueirense): {jogo['avai']} x {jogo['figueirense']}")

print('\nVitórias do Avaí:', vitorias_avai)
print('Vitórias do Figueirense:', vitorias_figueirense)
print('Empates:', empates)
print('Saldo de gols do Avaí:', saldo_avai)
print('Saldo de gols do Figueirense:', saldo_figueirense)
```

## Como interpretar o saldo de gols

Em cada partida, o saldo do Avaí recebe `gols do Avaí - gols do Figueirense`; o do Figueirense recebe a diferença inversa. Os saldos são acumulados nos cinco jogos, então podem ser positivos, negativos ou iguais a zero. As estatísticas são calculadas consultando os dados guardados nos dicionários da lista `jogos`.
