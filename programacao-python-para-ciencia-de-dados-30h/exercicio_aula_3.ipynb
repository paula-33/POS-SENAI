OBJETIVO: Desenvolva um algoritmo de uma função que receba o preço (em reais) por litro da Gasolina e do
Etanol.
O programa deverá retornar a melhor opção de combustível (GASOLINA ou ETANOL) de acordo com os
preços informados.
REQUISITOS:
✔ Considere que uma relação de 73% ou menos do preço do etanol em relação ao preço da gasolina,
favorece o uso do etanol. Se for acima de 73%, é mais viável a utilização da gasolina;
✔ Considere que o preço por litro da gasolina e do etanol devem estar entre R$ 0,01 e R$ 10,00 (0.01 e
10.00). Qualquer valor fora desse intervalo, o programa deverá informar “PREÇO INVÁLIDO”.


### Resolução

A função valida os dois preços antes de comparar. A relação é calculada dividindo o preço do etanol pelo da gasolina; se o resultado for menor ou igual a `0.73`, a função recomenda **ETANOL**. Caso contrário, recomenda **GASOLINA**.


```python
def melhor_combustivel(preco_etanol, preco_gasolina):
    # Os dois preços devem estar no intervalo permitido, inclusive.
    if not (0.01 <= preco_etanol <= 10.00 and 0.01 <= preco_gasolina <= 10.00):
        return 'PREÇO INVÁLIDO'

    # Etanol compensa quando custa até 73% do preço da gasolina.
    if preco_etanol / preco_gasolina <= 0.73:
        return 'ETANOL'
    return 'GASOLINA'

preco_etanol = float(input('Digite o preço do etanol por litro (R$): '))
preco_gasolina = float(input('Digite o preço da gasolina por litro (R$): '))

resultado = melhor_combustivel(preco_etanol, preco_gasolina)
print(resultado)
```
