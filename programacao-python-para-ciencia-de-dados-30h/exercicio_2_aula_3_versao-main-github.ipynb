# Exercício 2 — Lançamento de dados

Simule dez lançamentos de um dado e armazene-os em uma lista. Calcule a média, a quantidade de resultados acima dela, o maior valor e o menor valor, usando os elementos armazenados.

A solução não usa laços `for` como instruções: as compreensões de lista geram os lançamentos e selecionam os que ficam acima da média.


```python
import random

# Gera e armazena os 10 resultados (de 1 a 6).
lancamentos = [random.randint(1, 6) for _ in range(10)]

# Calcula a média a partir dos valores guardados.
media = sum(lancamentos) / len(lancamentos)

# Seleciona e conta somente os valores acima da média.
acima_da_media = [valor for valor in lancamentos if valor > media]
quantidade_acima_da_media = len(acima_da_media)

maior_valor = max(lancamentos)
menor_valor = min(lancamentos)

print('Valores dos lançamentos:', lancamentos)
print('Média dos valores:', media)
print('Quantidade de valores acima da média:', quantidade_acima_da_media)
print('Maior valor:', maior_valor)
print('Menor valor:', menor_valor)
```

## Como funciona

`sum` soma os valores e `len` conta quantos lançamentos foram armazenados. A compreensão `acima_da_media` filtra valores estritamente maiores que a média. `max` e `min` encontram os extremos da lista.
