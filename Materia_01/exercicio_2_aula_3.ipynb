# Exercício 2 — Lançamento de dados

Simule 10 lançamentos de um dado, armazenando cada resultado em uma lista. A partir dos valores guardados, calcule a média, quantos resultados ficaram acima da média, o maior valor e o menor valor.

## Estratégia

1. Gere dez números inteiros aleatórios entre 1 e 6 e guarde-os na lista `lancamentos`.
2. Calcule a média usando a soma dos elementos dividida pela quantidade de elementos.
3. Percorra a lista para contar os valores acima da média.
4. Encontre o maior e o menor valor da própria lista.

A cada execução, os lançamentos podem ser diferentes. O exemplo do enunciado `[1, 6, 2, 1, 6, 3, 3, 3, 4, 5]` tem média `3.4`, quatro valores acima da média, máximo `6` e mínimo `1`.


```python
import random

# 1. Gera 10 lançamentos
lancamentos = [random.randint(1, 6) for _ in range(10)]

# 2. Calcula a média
media = sum(lancamentos) / len(lancamentos)

# 3. Conta os valores acima da média
quantidade_acima_da_media = 0

for valor in lancamentos:
    if valor > media:
        quantidade_acima_da_media += 1

# 4. Maior e menor valor
maior_valor = max(lancamentos)
menor_valor = min(lancamentos)

print("Valores dos lançamentos:", lancamentos)
print("Média dos valores:", media)
print("Quantidade de valores acima da média:", quantidade_acima_da_media)
print("Maior valor:", maior_valor)
print("Menor valor:", menor_valor)
```
