# Exploração CNES — estrutura de diálise, competência 2024-12

**Objetivo:** descrever estabelecimentos com perfil complementar de diálise registrado no CNES em dezembro/2024, por município gestor cadastrado, sem interpretá-los como capacidade assistencial efetiva.

O arquivo ZIP tem cerca de 2,67 GB descompactado. O notebook lê apenas `tbDialise202412.csv`, `tbEstabelecimento202412.csv` e `tbMunicipio202412.csv` diretamente do ZIP, sem extraí-las para disco.


## Fonte e definição

Fonte local: `Dados/01_fontes_originais/CNES/BASE_DE_DADOS_CNES_202412.ZIP`, base CNES da competência dezembro/2024. Referência: [documentação CNES](https://cnes.datasus.gov.br/pages/downloads/documentacao.jsp).

A tabela `tbDialise` descreve salas e máquinas declaradas no cadastro. Uma linha não comprova funcionamento, atendimento SUS, vagas, produção ou capacidade disponível. A análise é exploratória e de competência única.


## 1. Localizar ZIP e inspecionar tabelas

```python
from pathlib import Path
from zipfile import ZipFile
import csv, io

CANDIDATOS=[Path("../Dados/01_fontes_originais/CNES/BASE_DE_DADOS_CNES_202412.ZIP"),Path("../Dados/BASE_DE_DADOS_CNES_202412.ZIP"),Path("estatistica-aplicada-a-analise-de-dados-40h/Trabalho-final/Dados/BASE_DE_DADOS_CNES_202412.ZIP")]
ZIP_PATH=next((p for p in CANDIDATOS if p.exists()),CANDIDATOS[0])
assert ZIP_PATH.exists(),f"ZIP CNES não encontrado: {ZIP_PATH}"
MEMBROS={
 "dialise":"tbDialise202412.csv",
 "estabelecimento":"tbEstabelecimento202412.csv",
 "municipio":"tbMunicipio202412.csv",
}
with ZipFile(ZIP_PATH) as z:
    disponiveis=set(z.namelist())
    for apelido,nome in MEMBROS.items():
        assert nome in disponiveis, f"Membro ausente no ZIP: {nome}"
        print(apelido,nome,f"{z.getinfo(nome).file_size/1_000_000:.1f} MB descompactado")
```

## 2. Leitura em streaming de registros selecionados

```python
def iterar_csv_cnes(nome_membro):
    """Lê linhas latin-1 separadas por ponto e vírgula diretamente do ZIP."""
    with ZipFile(ZIP_PATH) as z:
        with z.open(nome_membro) as binario:
            texto=io.TextIOWrapper(binario,encoding="latin-1",newline="")
            leitor=csv.DictReader(texto,delimiter=";")
            for linha in leitor: yield linha

# Carrega os registros necessários para ligar o perfil de diálise ao estabelecimento e ao município gestor.
dialise=list(iterar_csv_cnes(MEMBROS["dialise"]))
assert dialise, "Tabela tbDialise sem registros."
print("Registros tbDialise:",len(dialise))
print("Campos:",list(dialise[0].keys()))

estab={r["CO_UNIDADE"]:r for r in iterar_csv_cnes(MEMBROS["estabelecimento"])}
municipios={r["CO_MUNICIPIO"]:r for r in iterar_csv_cnes(MEMBROS["municipio"])}
print("Estabelecimentos indexados:",len(estab))
print("Municípios indexados:",len(municipios))
print("Campos município:",list(next(iter(municipios.values())).keys()))
```

## 3. Verificar chaves e montar tabela analítica

```python
chaves=[r["CO_UNIDADE"] for r in dialise]
assert len(chaves)==len(set(chaves)), "CO_UNIDADE repetido em tbDialise."
sem_estab=[c for c in chaves if c not in estab]
assert not sem_estab, f"Unidades de diálise sem correspondência em tbEstabelecimento: {len(sem_estab)}"

# Campos numéricos de estrutura: vazio continua ausente, não é convertido em zero.
campos_qt=["QT_SALA_HBSAG_POS","QT_SALA_HBSAG_NEG","QT_SALA_DPI","QT_SALA_DPAC","QT_SALA_REAG_POS","QT_SALA_REAG_NEG","QT_SALA_REHCV","NU_MAQH_PROP","NU_MAQH_OUTR"]
def numero_ou_none(x):
    x=(x or "").strip()
    return int(x) if x else None

analitica=[]
for d in dialise:
    e=estab[d["CO_UNIDADE"]]
    codigo_municipio_gestor=e.get("CO_MUNICIPIO_GESTOR","").strip()
    m=municipios.get(codigo_municipio_gestor,{})
    rec={
        "co_unidade":d["CO_UNIDADE"],
        "cnes":e.get("CO_CNES",""),
        "nome":e.get("NO_FANTASIA","").strip(),
        "codigo_municipio_gestor":codigo_municipio_gestor,
        "municipio":m.get("NO_MUNICIPIO",""),
        "sigla_uf":m.get("CO_SIGLA_ESTADO",""),
        "atendimento_sus_avaliado":False,  # Este notebook não avalia atendimento SUS.
    }
    for campo in campos_qt: rec[campo]=numero_ou_none(d.get(campo))
    analitica.append(rec)

print("Perfis de diálise ligados a estabelecimento e município gestor:",len(analitica))
print("Sem sigla de UF identificada:",sum(not r["sigla_uf"] for r in analitica))
```

## 4. Resumos descritivos de estrutura cadastrada

```python
por_uf={}
for r in analitica:
    uf=r["sigla_uf"] or "UF não identificada"
    por_uf[uf]=por_uf.get(uf,0)+1
print("Estabelecimentos com perfil de diálise por UF do município gestor:")
for uf,n in sorted(por_uf.items(),key=lambda x:x[1],reverse=True): print(uf,n)

for campo in ["NU_MAQH_PROP","NU_MAQH_OUTR"]:
    conhecidos=[r[campo] for r in analitica if r[campo] is not None]
    print(f"\n{campo}: registros não ausentes={len(conhecidos)}, soma={sum(conhecidos):,}, mediana={sorted(conhecidos)[len(conhecidos)//2] if conhecidos else None}".replace(",","."))

# Triagem de perfis com ao menos uma máquina de hemodiálise declarada no cadastro.
com_maquina=sum((r["NU_MAQH_PROP"] or 0)+(r["NU_MAQH_OUTR"] or 0)>0 for r in analitica)
print("Unidades com ao menos uma máquina declarada:",com_maquina,"de",len(analitica))
```

### Interpretação, limitações e próximos passos

- A contagem representa perfis complementares em `tbDialise` na competência 2024-12; não é série temporal, pacientes nem capacidade assistencial efetiva.
- A agregação territorial usa o município gestor cadastrado. A equivalência com a localização física deve ser confirmada na documentação CNES antes de publicação.
- Este notebook não avalia atendimento SUS. Para esse recorte, usar os CSVs derivados que relacionam serviço/classificação e os códigos de atendimento SUS.
- Salas e máquinas são valores declarados no cadastro; não comprovam instalação, funcionamento, exclusividade SUS, vagas ou produção.
- Os dados CNES ficam como referência contextual e não devem compor o indicador principal de oferta real do estudo.
